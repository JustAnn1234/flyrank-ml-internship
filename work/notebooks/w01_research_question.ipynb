## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Search Question
> *"Which published web pages exhibit significant performance decay or untapped traffic demand, and should be prioritized first by the content team for a refresh or rewrite?"*

### Core Framing Dimensions
* **Unit of Analysis (Grain):** A single unique content item (`content_id` / `content_hash_id`) aggregated over a 90-day feature window.
* **Model Output:** A continuous **Priority Refresh Score** (0–100) and probability of decay, paired with human-readable reason codes (e.g., `declining_with_demand`, `stale_visible_page`).
* **Decision:** Deciding which top 5%–10% of declining or high-demand content items should be assigned to editorial/growth teams for updates next month.
* **Action:** The content team reviews the flagged pages, updates outdated facts, expands thin sections, or restructures headings/meta descriptions.
* **Cost of a Wrong Call:**
  * **False Positive (High priority assigned to a non-performing or irrecoverable page):** Wasted editorial budget and reviewer time spent updating a page that yields zero traffic recovery.
  * **False Negative (High-potential declining page missed):** Organic search traffic continues to decay, causing a persistent loss of site traffic to competitors.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
import numpy as np

# List of relative local paths to try, followed by a raw GitHub fallback
data_url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

paths_to_try = [
    '../data/raw/content_refresh_anonymized.csv',
    'data/raw/content_refresh_anonymized.csv',
    '../../data/raw/content_refresh_anonymized.csv',
    data_url
]

df = None
for path in paths_to_try:
    try:
        df = pd.read_csv(path)
        print(f"Successfully loaded data from: {path}")
        break
    except Exception:
        continue

if df is None:
    raise FileNotFoundError("Could not load content_refresh_anonymized.csv from local paths or GitHub.")

# --- Number 1: Total Scale & Scope ---
total_rows = len(df)

# --- Number 2: Opportunity Volume (Declining with Demand) ---
declining_pages = df[(df['impressions_90d'] >= 100) & (df['trend_direction'] == 'down')]
pct_declining = (len(declining_pages) / total_rows) * 100

# --- Number 3: AI Traffic Footprint ---
ai_active_pages = df[df['ai_sessions_90d'] > 0]
pct_ai_active = (len(ai_active_pages) / total_rows) * 100

print("\n=== STARTER DATASET SUMMARY METRICS ===")
print(f"1. Total Content Items Analyzed: {total_rows:,} rows")
print(f"2. Refresh Candidates (Impressions >= 100 & Trend == 'down'): {len(declining_pages):,} pages ({pct_declining:.1f}%)")
print(f"3. AI Traffic Exposure (Pages with >0 AI Sessions): {len(ai_active_pages):,} pages ({pct_ai_active:.1f}%)")
```

### Why these numbers justify the lane:
1. **Massive Opportunity Volume:** Exactly **13,152 pages (43.8%)** in the dataset maintain active search demand (`impressions_90d >= 100`) yet show a downward trend (`trend_direction == 'down'`). Manually auditing over 13,000 declining pages is unfeasible for a content team, proving the clear need for an automated priority ranking system.
2. **AI Traffic Analysis Integration:** A total of **1,930 pages (6.4%)** display active AI referral sessions (`ai_sessions_90d > 0`). This provides a strong sample size to analyze how traditional search decay interacts with AI engine traffic.
